```python
import json
import pandas as pd
```

```python
file_path = 'api_data_files/international_box_player_season.json'
with open(file_path, 'r') as f:
    json_data = json.load(f)
inter_box = pd.read_json(file_path)
nba_box = pd.read_json('api_data_files/nba_box_player_season.json')
player = pd.read_json('api_data_files/player.json')
```

```python
# inter_box.info()
inter_box['first_name'] = inter_box['first_name'].str.title()
inter_box['last_name'] = inter_box['last_name'].str.title()
inter_box
```

```python
# nba_box.info()
nba_box['first_name'] = nba_box['first_name'].str.title()
nba_box['last_name'] = nba_box['last_name'].str.title()
nba_box
```

```python
player.head()
player['first_name'] = player['first_name'].str.title()
player['last_name'] = player['last_name'].str.title()
player.head()
```

```python
combined = pd.concat([inter_box, nba_box])
combined.info()
```

```python
# join birth_date to first and last name
# notice capital vs non capital letters
# join data
merged_df = pd.merge(combined, player, on=['first_name', 'last_name'], how='left')
merged_df.info()
merged_df.to_csv('merged.csv', index=False)
```

```python
# normalize first/last names into a single key to reduce mismatches
for df in [inter_box, nba_box, player]:
    df['first_name_norm'] = df['first_name'].astype(str).str.strip().str.lower()
    df['last_name_norm']  = df['last_name'].astype(str).str.strip().str.lower()
    df['full_name_norm']  = df['first_name_norm'] + '|' + df['last_name_norm']  # safe composite key
```

```python
# boolean mask per row, then counts
inter_box['in_nba'] = inter_box['full_name_norm'].isin(nba_box['full_name_norm'])
inter_box['in_nba'].value_counts(dropna=False)   # shows how many True/False/NaN
inter_box['in_nba'].sum()                        # number of rows in inter_box that match nba_box
len(inter_box)                                   # total rows in inter_box
```

```python
inter_names = set(inter_box['full_name_norm'].dropna().unique())
nba_names   = set(nba_box['full_name_norm'].dropna().unique())
player_names= set(player['full_name_norm'].dropna().unique())

common_inter_nba = inter_names & nba_names
len(common_inter_nba), "unique full names in both inter_box and nba_box"

# also for all three:
common_all_three = inter_names & nba_names & player_names
len(common_all_three), "unique full names present in all three datasets"
```
